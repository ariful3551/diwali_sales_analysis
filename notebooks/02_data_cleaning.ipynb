##### Data Load

```python
import numpy as np
import pandas as pd

df = pd.read_csv("../data/raw/diwali_sales_data.csv", encoding="latin-1")
df_clean = df.copy()
```

##### Data Clean

```python
# Remove the last two empty columns (Status, unnamed1)
# Rename columns: 'Cust_name' -> 'Customer_name', 'Age Group' -> 'Age_Group'
# Change the data type of the 'Age Group' column
# Change the data type of the 'Amount' column
# Remove null value
# Check `Age` against `Age_Group`
```

```python
df_clean = df_clean.drop(columns= ["Status", "unnamed1"])
df_clean.shape
```

```python
df_clean.rename(columns={"Cust_name" : "Customer_Name", "Age Group" : "Age_Group"}, inplace= True)
df_clean.columns
```

```python
df_clean.dropna(inplace=True)
df_clean.isnull().sum()
```

```python
df_clean.shape
```

```python
(df_clean["Amount"] % 1 != 0).sum() 

df_clean["Age_Group"] = df_clean["Age_Group"].astype("category")
df_clean.dtypes
```

```python
df_clean["Amount"] = df_clean["Amount"].round().astype(int)
df_clean.dtypes
```

```python
df_clean.groupby("Age_Group")["Age"].agg(["min", "max", "count"])
```
